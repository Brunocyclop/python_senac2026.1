# set (Conjunto)

set (Conjunto)

O que é: É uma coleção não ordenada e mutável de elementos únicos. É uma estrutura de dados utilizada para armazenar valores sem repetições em Python.

Características:

Não ordenado: Os elementos não possuem índices nem uma posição fixa. A ordem de exibição não deve ser considerada garantida.

Mutável: Podemos adicionar ou remover elementos após a criação do conjunto.

Não permite duplicatas: Elementos repetidos são automaticamente eliminados.

Heterogêneo: Pode armazenar diferentes tipos de dados, como números e strings, desde que os elementos sejam hashable (tenham um valor de hash estável).

Sem indexação: Não podemos acessar elementos por índices, como fazemos em listas.

Operações matemáticas: Permite realizar união, interseção, diferença e diferença simétrica entre conjuntos.

Quando usar:

Quando você precisa armazenar elementos únicos, sem repetições.

Quando deseja eliminar valores duplicados de uma lista.

Quando precisa verificar rapidamente se um elemento pertence a uma coleção.

Quando precisa comparar grupos de dados e encontrar elementos em comum ou diferentes.

Para implementar operações matemáticas entre conjuntos.




# exemplos

```python


numeros = {10, 20, 30, 20, 10}

print(numeros)

Resultado possível:

{10, 20, 30}
```

# Os valores repetidos são eliminados automaticamente.



- **Principais métodos e operações:**
  - `add()`: adiciona um elemento.
  - `remove()`: remove um elemento; gera `KeyError` se ele não existir.
  - `discard()`: remove um elemento, sem gerar erro se ele não existir.
  - `pop()`: remove e retorna um elemento arbitrário.
  - `clear()`: remove todos os elementos.
  - `union()` ou `|`: combina os elementos de dois conjuntos.
  - `intersection()` ou `&`: retorna os elementos comuns.
  - `difference()` ou `-`: retorna os elementos que estão no primeiro conjunto, mas não no segundo.
  - `symmetric_difference()` ou `^`: retorna os elementos presentes em apenas um dos conjuntos.

- **Observação importante:** Para criar um conjunto vazio, utilize `set()`. O uso de `{}` cria um dicionário vazio (`dict`), e não um conjunto.

Em Python, o set (conjunto) é uma estrutura de dados utilizada para armazenar elementos únicos, eliminar duplicatas e realizar operações entre grupos de informações.

# criando um conjunto 

```python
frutas = {"maçã", "banana", "laranja"}

print(frutas)
```

# Resultado possível:

{'maçã', 'banana', 'laranja'}

- As chaves {} delimitam os elementos do conjunto.
- Os elementos não possuem índices.
- A ordem de exibição não é garantida.
2. Eliminando elementos duplicados
Um dos usos mais comuns do set é remover valores repetidos.

```python
numeros = {10, 20, 10, 30, 20, 40}

print(numeros)
```

# Resultado:

```python
{10, 20, 30, 40}
```
