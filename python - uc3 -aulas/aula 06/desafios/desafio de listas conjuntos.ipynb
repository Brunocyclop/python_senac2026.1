# Guia de conjuntos em Python

O que é
Um conjunto (set) é uma coleção não ordenada de elementos únicos. Em Python, criamos conjuntos com chaves ({}) ou com a função set(). Eles são úteis quando a ordem não importa, mas a ausência de repetições importa.

thon
# Conjunto de categorias de produtos de uma loja.
categorias = {"eletrônicos", "livros", "papelaria"}

# Conjunto vazio precisa usar set(), pois {} cria um dicionário.
categorias_vazias = set()

# Características

Sem duplicatas: se você tentar adicionar um item que já existe, nada muda.

Não ordenados: não há índices como em listas; você não pode fazer categorias[0].

Mutáveis: é possível adicionar e remover itens em um set.

Elementos devem ser hasheáveis: textos, números e tuplas podem ser itens; listas e dicionários não.

Verificação rápida: testar se um item está presente com in costuma ser eficiente.


python
categorias = {"eletrônicos", "livros", "papelaria"}

# Tentativa de adicionar uma categoria repetida.
categorias.add("livros")

# O conjunto continua com apenas uma ocorrência de "livros".
print(categorias)

# Quando utilizar ?

Use conjuntos para representar grupos sem repetição, como categorias, e-mails cadastrados, permissões de usuário ou tags de um produto. Também são excelentes para remover duplicatas de uma lista e comparar grupos de dados.

```python
python
# Remove categorias duplicadas de uma lista.
categorias_recebidas = ["livros", "papelaria", "livros", "eletrônicos"]

categorias_unicas = set(categorias_recebidas)

print(categorias_unicas)
```

# Principais métodos

Os exemplos usam duas listas de categorias de uma loja: as categorias cadastradas e as categorias disponíveis em estoque. Cada exemplo parte dos mesmos conjuntos iniciais. Os métodos add(), remove(), union(), intersection() e difference() fazem parte do tipo set da biblioteca padrão de Python.

add(item)
Adiciona um elemento ao conjunto. Se ele já existir, o conjunto permanece igual.

```python
python
categorias = {"eletrônicos", "livros"}

# Adiciona uma nova categoria.
categorias.add("papelaria")

print(categorias)  # {'eletrônicos', 'livros', 'papelaria'}
```

# update(iteravel)
Adiciona vários elementos de uma lista, tupla ou outro conjunto.

python


```python
python
categorias = {"eletrônicos", "livros"}

# Acrescenta várias categorias de uma vez.
categorias.update(["papelaria", "brinquedos"])

print(categorias)  # {'eletrônicos', 'livros', 'papelaria', 'brinquedos'}
```

# remove(item)
Remove um elemento. Se o elemento não existir, gera um erro KeyError.

```python
python
categorias = {"eletrônicos", "livros", "papelaria"}

# Remove uma categoria existente.
categorias.remove("livros")

print(categorias)  # {'eletrônicos', 'papelaria'}
```

# discard(item)
Remove um elemento apenas se ele existir. Se não existir, não gera erro.

```python
python
categorias = {"eletrônicos", "livros", "papelaria"}

# Remove "brinquedos", mesmo que ele não esteja no conjunto.
categorias.discard("brinquedos")

print(categorias)  # {'eletrônicos', 'livros', 'papelaria'}
```

# union(outro_conjunto)
Cria um novo conjunto com todos os elementos dos dois conjuntos, sem repetições. O operador equivalente é |.




```python
categorias_cadastradas = {"eletrônicos", "livros", "papelaria"}
categorias_em_estoque = {"livros", "brinquedos"}

# Junta as duas categorias.
todas_categorias = categorias_cadastradas.union(categorias_em_estoque)

print(todas_categorias)
# {'eletrônicos', 'livros', 'papelaria', 'brinquedos'}
```

# intersection(outro_conjunto)
Cria um novo conjunto apenas com os elementos presentes nos dois conjuntos. O operador equivalente é &.

```python
categorias_cadastradas = {"eletrônicos", "livros", "papelaria"}
categorias_em_estoque = {"livros", "brinquedos"}

# Categorias cadastradas que também estão em estoque.
categorias_disponiveis = categorias_cadastradas.intersection(categorias_em_estoque)

print(categorias_disponiveis)  # {'livros'}
```

# difference(outro_conjunto)
Cria um novo conjunto com os elementos do primeiro que não estão no segundo. O operador equivalente é -.

```python
categorias_cadastradas = {"eletrônicos", "livros", "papelaria"}
categorias_em_estoque = {"livros", "brinquedos"}

# Categorias cadastradas que não estão em estoque.
categorias_sem_estoque = categorias_cadastradas.difference(categorias_em_estoque)

print(categorias_sem_estoque)  # {'eletrônicos', 'papelaria'}
```

# symmetric_difference(outro_conjunto)
Cria um novo conjunto com os elementos que estão em apenas um dos conjuntos, mas não em ambos. O operador equivalente é ^.




```python
categorias_cadastradas = {"eletrônicos", "livros", "papelaria"}
categorias_em_estoque = {"livros", "brinquedos"}

# Categorias que estão em apenas um dos dois conjuntos.
categorias_exclusivas = categorias_cadastradas.symmetric_difference(categorias_em_estoque)

print(categorias_exclusivas)  # {'eletrônicos', 'papelaria', 'brinquedos'}
```

# issubset(outro_conjunto)
Verifica se todos os elementos de um conjunto estão presentes em outro conjunto.

```python
categorias_cadastradas = {"eletrônicos", "livros", "papelaria"}
categorias_promocionais = {"livros", "papelaria"}

# Verifica se as promoções fazem parte das categorias cadastradas.
promocoes_validas = categorias_promocionais.issubset(categorias_cadastradas)

print(promocoes_validas)  # True
```

# clear()
Remove todos os elementos do conjunto.

```python
categorias = {"eletrônicos", "livros", "papelaria"}

# Esvazia o conjunto.
categorias.clear()

print(categorias)  # set()
```
