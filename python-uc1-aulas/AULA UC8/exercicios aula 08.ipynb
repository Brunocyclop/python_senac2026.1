# Exercício 3: Função par_ou_impar
Objetivo: Verifique se um número é par ou ímpar.
Passos:
Defina a função par_ou_impar com um parâmetro n.
Use a operação de módulo (%) para verificar se não é divisível por 2.
Retornar a string "Par" se for divisível ou "Ímpar" caso contrário.
Chamar a função e exibir o resultado.


```python
def par_ou_impar(n):
    if n % 2 == 0:
        return "Par"
    else:
        return "Ímpar"
# Exemplo de chamada
print("O número 4 é", par_ou_impar(4))
print("O número 7 é", par_ou_impar(7))
```

# 2️⃣ Crie um programa Python que use a função main()para calcular e imprimir a soma de dois números fornecidos pelo usuário.


```python
def main():
   num1 = float(input("Digite o primeiro número: "))
   num2 = float(input("Digite o segundo número: "))
   soma = num1 + num2
   print(f"A soma é: {soma}")

if __name__ == "__main__":
   main()
```

# 3️⃣ Escreva um programa Python que use a função main()para chamar outras funções, como soma()e subtracao(), e exiba os resultados.


```python
def soma(a, b):
   return a + b

def subtracao(a, b):
   return a - b

def main():
   resultado_soma = soma(30, 7)
   resultado_subtracao = subtracao(30, 7)
   print(f"Soma: {resultado_soma}")
   print(f"Subtração: {resultado_subtracao}")

if __name__ == "__main__":
   main()
```

# 5️⃣ Escreva um programa Python que use uma função main()para calcular o fatorial de um número fornecido pelo usuário.

```python
def fatorial(n):
   if n == 0 or n == 1:
       return 1
   else:
       return n * fatorial(n - 1)

def main():
   num = int(input("Digite um número: "))
   print(f"O fatorial de {num} é {fatorial(num)}")

if __name__ == "__main__":
   main()
```
