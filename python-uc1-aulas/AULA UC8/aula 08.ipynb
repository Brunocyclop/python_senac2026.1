```python
def soma(a,b):
    soma=a+b
    return soma
```

```python
def soma_2(n1,n2):
    soma=n1+n2
    print(soma)
    
```

```python
def potencia (numero, expoente):
    resultado = numero**expoente
    return resultado
```

```python
numero=2
expoente=8

print(potencia(numero,expoente))
```

```python
n2=5
exp=4
resultado=potencia(n2,exp)
resultado2=potencia(exp,n2)
```

```python
print(f"{resultado}<>{resultado2}")
```

# passa de arguments


```python
def saudacao(display="sem nome"):
    print (f"olá,  {display}!!!")
```

```python
saudacao("bruno")
```

```python
saudacao()
```

# passagem de argumento arbirtrarios

```python
def saudacoes (*nomes):
    #print (nomes)
    for nome in nomes:
        saudacao(nome)
    
```

```python
saudacoes ("Bruno" ,"patricio" , "pereira")
```
