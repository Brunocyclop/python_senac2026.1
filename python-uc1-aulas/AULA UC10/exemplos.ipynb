# 6️⃣: Encontrar o maior e o menor valor em um vetor
📜 Explicação:

Passo 1: Criamos o vetor com alguns números.
Passo 2: max(vetor) retorna o maior número e min(vetor) o menor.
Passo 3: Exibimos os valores encontrados.

```python
# Passo 1: Definir o vetor.
vetor = [3, 1, 2, 10, 4]
# Passo 2: Utilizar as funções max() e min() para encontrar os valores.
maior = max(vetor)
menor = min(vetor)
# Passo 3: Imprimir os resultados.
print("Maior valor:", maior)
print("Menor valor:", menor)
```

# 7️⃣: Inverter a ordem dos elementos de um vetor sem usar reverse()
📜 Explicação:

Passo 1: Definimos um vetor simples.
Passo 2: O slicing [::-1] cria uma nova lista com os elementos na ordem inversa.
Passo 3: Imprimimos o resultado.

```python
# Passo 1: Definir o vetor.
vetor = [1, 2, 3, 4, 5]
# Passo 2: Inverter o vetor utilizando slicing.
vetor_invertido = vetor[::-1]
# Passo 3: Imprimir o vetor invertido.
print("Vetor invertido:", vetor_invertido)
```

```python

vetor = [1, 2, 4, 5, 6]
vetor_invertido = vetor[::-1]
print("Vetor invertido:", vetor_invertido)
```
