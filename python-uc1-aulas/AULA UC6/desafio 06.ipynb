# Imprimir Números pares e impares
🔖 Elabore um código para imprimir todos os números pares e ímpares de 1 a 20 usando for

```python
print(f"os numeros pares de 1 a 20 sao:")
for numero in range (1, 21):
    if (numero % 2 ==0) :
        print(f"{numero:2d}",end=",")
```

```python
print (f" os numeros impares de 1 a 20 sao:")
for numero in range (1, 21):
    if (numero % 2 == 1) :
        print(f"{numero:2d}",end=",")
```

```python
pares=[]
impares=[]
for numero in range (1, 21):
    if (numero % 2 == 0):
        pares.append(numero)
print (f"numeros pares  :{pares}")
print (f"numeros impares  :{impares}")
```

# Imprimir de 10 até 1
🔖 Elabore um código para imprimir de 10 até 1 de forma decrescente, usando while

```python
numero = 10
while numero > 0 :
    print (f"{numero}",end=",")
    numero=numero-1
```

# Imprimir a soma dos números
🔖 Elabora um código para imprimir a soma dos números de 1 a 100 usando for

```python
soma=0
for numero in range (1, 101):
    soma = soma + numero
    print(f"o valor da soma de todos os numeros de 1 até 100 é {soma}")
```
